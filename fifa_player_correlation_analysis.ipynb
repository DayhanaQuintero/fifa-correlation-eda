### Crecimiento por GÉNERO

```python
#Aunque el dataset contiene registros hasta el año 2020, los años más recientes 
# presentan un número muy reducido de observaciones (por ejemplo, solo 3 juegos en 2017 y 1 en 2020).
#Para garantizar un análisis de crecimiento representativo, se seleccionaron los últimos cuatro años con volumen significativo de datos (2013–2016), 
#permitiendo obtener conclusiones válidas sobre tendencias de mercado.

genre_year = (df_4.groupby(["Year", "Genre"])["Global_Sales"].sum().reset_index())
```

```python
genre_pivot = genre_year.pivot(index="Genre",columns="Year",values="Global_Sales").fillna(0)
```

```python
genre_pivot["Crecimiento"] = genre_pivot[2016] - genre_pivot[2013]
genre_growth = genre_pivot.sort_values("Crecimiento", ascending=False)

print("Mayor crecimiento género:")
print(genre_growth.head(1))

print("\nMenor crecimiento género:")
print(genre_growth.tail(1))
```

### Crecimiento por PLATAFORMA

```python
platform_year = (df_4.groupby(["Year", "Platform"])["Global_Sales"].sum().reset_index())

platform_pivot = platform_year.pivot(index="Platform",columns="Year",values="Global_Sales").fillna(0)

platform_pivot["Crecimiento"] = platform_pivot[2016] - platform_pivot[2013]

platform_growth = platform_pivot.sort_values("Crecimiento", ascending=False)

print("Mayor crecimiento plataforma:")
print(platform_growth.head(1))

print("\nMenor crecimiento plataforma:")
print(platform_growth.tail(1))
```

### INSIGHTS ADICIONALES

```python
#INSIGHT 1 — Géneros con mayor crecimiento en una región
#El mercado no crece igual en todas las regiones.
genre_region = (df_4.groupby(["Year", "Genre"])["NA_Sales"].sum().reset_index())

genre_pivot_na = genre_region.pivot(index="Genre",columns="Year",values="NA_Sales").fillna(0)

genre_pivot_na["Crecimiento_NA"] = genre_pivot_na[2016] - genre_pivot_na[2013]

genre_pivot_na.sort_values("Crecimiento_NA", ascending=False).head(5)
```

```python
#INSIGHT 2 — Plataformas con mayor venta promedio por juego
#No solo importa vender mucho, importa cuánto vende cada juego en promedio
df_2016 = df_4[df_4["Year"] == 2016]

platform_avg = (df_2016.groupby("Platform").agg(Juegos=("Name", "count"), Ventas=("Global_Sales", "sum")))

platform_avg["Venta_Promedio"] = (platform_avg["Ventas"] / platform_avg["Juegos"])*100

platform_avg.sort_values("Venta_Promedio", ascending=False).head(5)
```

### Gráficos presentación

```python
import pandas as pd
import numpy as np
import plotly.express as px

df = pd.read_csv("vgsales.csv")
df.columns = df.columns.str.strip()

df["Year"] = pd.to_numeric(df["Year"], errors="coerce")

y0, y1 = 2013, 2016
df_4 = df[df["Year"].between(y0, y1)].copy()

sales_cols = ["NA_Sales","EU_Sales","JP_Sales","Other_Sales","Global_Sales"]
sales_cols
```

```python
# SLIDE 2: Ventas totales por año (Global)

ventas_por_anio = (df_4.groupby("Year")[sales_cols].sum().reset_index().sort_values("Year"))

fig2 = px.bar( ventas_por_anio,x="Year", y="Global_Sales",title=f"Ventas globales totales por año ({y0}–{y1})",
              text_auto=".2f", width=1000,height=500, color="Year",color_discrete_sequence=px.colors.qualitative.Pastel)
fig2.show()
```

```python
# SLIDE 3: Ventas por género por año (líneas)
genre_year = (df_4.groupby(["Year","Genre"])["Global_Sales"].sum().reset_index())

top_genres = (df_4.groupby("Genre")["Global_Sales"].sum().sort_values(ascending=False).head(6).index)

genre_year_top = genre_year[genre_year["Genre"].isin(top_genres)].sort_values(["Year","Genre"])

fig3 = px.line(genre_year_top,x="Year",y="Global_Sales",color="Genre",markers=True,
               title=f"Evolución de ventas globales por género (Top 6) ({y0}–{y1})",width=1000,height=500)
fig3.show()
```

```python
# SLIDE 4: Crecimiento por plataforma (2016 - 2013)
platform_year = (df_4.groupby(["Year","Platform"])["Global_Sales"].sum() .reset_index())

platform_pivot = (platform_year.pivot(index="Platform", columns="Year", values="Global_Sales").fillna(0))

for y in [y0, y1]:
    if y not in platform_pivot.columns:
        platform_pivot[y] = 0

platform_pivot["Crecimiento"] = platform_pivot[y1] - platform_pivot[y0]

top_plat = platform_pivot["Crecimiento"].abs().sort_values(ascending=False).head(10).index
platform_growth_plot = platform_pivot.loc[top_plat, ["Crecimiento"]].reset_index().sort_values("Crecimiento", ascending=False)

fig4 = px.bar(platform_growth_plot,x="Platform",y="Crecimiento",color="Platform",
              title=f"Crecimiento de ventas globales por plataforma ({y1} – {y0}) | Top 10",text_auto=".2f",width=1000,height=500)

fig4.update_layout(showlegend=False)
fig4.show()
```

```python
# SLIDE 5: Juegos líderes del mercado (Top juego 2013 vs Top juego 2016)
df_2013 = df_4[df_4["Year"] == y0]
df_2016 = df_4[df_4["Year"] == y1]

top_game_2013 = df_2013.loc[df_2013["Global_Sales"].idxmax(), ["Year","Name","Platform","Genre","Global_Sales"]]
top_game_2016 = df_2016.loc[df_2016["Global_Sales"].idxmax(), ["Year","Name","Platform","Genre","Global_Sales"]]

top_games_compare = pd.DataFrame([top_game_2013, top_game_2016])
top_games_compare["Label"] = top_games_compare["Year"].astype(int).astype(str) + " — " + top_games_compare["Name"]

fig5 = px.bar(top_games_compare,x="Label",y="Global_Sales",color="Year",title=f"Cambio en el videojuego más vendido (Global) {y0} vs {y1}",
    text_auto=".2f",width=1000,height=500)
fig5.update_xaxes(title="")
fig5.show()
```

```python
# SLIDE 6 (Insight 1): Crecimiento de géneros por región (NA) (2016 - 2013)
genre_na_year = (df_4.groupby(["Year","Genre"])["NA_Sales"].sum().reset_index())

genre_na_pivot = (genre_na_year.pivot(index="Genre", columns="Year", values="NA_Sales").fillna(0))

for y in [y0, y1]:
    if y not in genre_na_pivot.columns:
        genre_na_pivot[y] = 0

genre_na_pivot["Crecimiento_NA"] = genre_na_pivot[y1] - genre_na_pivot[y0]

top_gen_na = genre_na_pivot["Crecimiento_NA"].abs().sort_values(ascending=False).head(8).index
genre_na_plot = genre_na_pivot.loc[top_gen_na, ["Crecimiento_NA"]].reset_index().sort_values("Crecimiento_NA", ascending=False)

fig6a = px.bar(genre_na_plot,x="Genre",y="Crecimiento_NA",color="Genre",title=f"Crecimiento de ventas en NA por género ({y1} – {y0}) | Top 8",
    text_auto=".2f",width=1000,height=500)
fig6a.update_layout(showlegend=False)
fig6a.show()
```

```python
# SLIDE 6 (Insight 2): Eficiencia (venta promedio por título) por plataforma en 2016
df_last = df[df["Year"] == y1].copy()

plat_eff = (df_last.groupby("Platform").agg(Juegos=("Name","count"), Ventas_Global=("Global_Sales","sum")).reset_index())

plat_eff["Venta_Promedio"] = plat_eff["Ventas_Global"] / plat_eff["Juegos"]

plat_eff_filtered = (plat_eff[plat_eff["Juegos"] >= 10].sort_values("Venta_Promedio", ascending=False).head(10))

fig6b = px.bar(plat_eff_filtered,x="Platform",y="Venta_Promedio",color="Platform",
    title=f"Venta promedio por videojuego (Global) por plataforma en {y1} | Top 10 (>=10 juegos)",text_auto=".2f",
    width=1000,height=500)
fig6b.update_layout(showlegend=False)
fig6b.show()
```
